```python
df= pd.read_csv("car-sales.csv")
```

```python
df.dtypes
```

```python
df['Price'] = (
    df['Price']
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(float)
    .astype(int)
)
```

```python
df['sale_Date'] = pd.date_range("1/1/2022", periods=len(df))
```

```python
df["total-sales"] = df['Price'].cumsum()
```

```python
df
```

```python
plt.plot( df['sale_Date'],df['total-sales']);
```

```python
df.plot(x="Make", y="Price", kind="bar",xlabel="Car Maker names", ylabel="Price of each car")
```

```python
df.plot(x="Make", y="Price", kind="scatter")
```

```python
plt.pie(x=df['Price'], labels=df['Make'],autopct='%1.1f%%')
```

```python
df= pd.read_csv("heart-disease.csv")
df
```

```python
plt.hist(df["age"], bins= 20);
```

```python
df.plot.hist(figsize=(10,20),subplots=True);
```

```python
df
```

```python
df_over50 = df[df['age']> 50]
```

```python
df_over50.plot(kind='scatter',
               x='age',
               y='chol',
               c='target')
```

```python
# OO method
fig, ax = plt.subplots(figsize=(10,10))
df_over50.plot(kind="scatter",x="age", y="chol",c="target",ax=ax)
```

```python
fig, ax = plt.subplots(figsize=(15,10))
scatter = ax.scatter(x=df_over50['age'], y=df_over50['chol'],c=df_over50['target'])
```

```python

fig, (ax1,ax2) = plt.subplots(nrows=2,ncols=1,figsize=(10,10))
scatter = ax1.scatter(x=df_over50['age'], y=df_over50['thal'],c=df_over50['target'],cmap="winter")
scatter = ax2.scatter(x=df_over50['age'], y=df_over50['chol'],c=df_over50['target'],cmap="winter")
ax2.axhline(df_over50['chol'].mean(),linestyle='--',color="r")
```
