```python
np.random.seed(seed=10)
mat1= np.random.randint(1,30,(3,5))
mat2= np.random.randint(1,30,(5,3))
```

```python
mat1
```

```python
mat2
```

```python
mat1*mat2
```

```python
#Dot product
np.dot(mat1, mat2)
```

```python
mat=np.array([[2,7,1],
              [9,4,16],
              [11,14,18],
              [13,13,16],
              [15,18,9]])
```

```python
Price=np.array([[10],[8],[12]])
Price
```

```python
total_sales=np.dot(mat,Price)
```

```python
total_sales
```

```python
import pandas as pd
```

```python
Sales = pd.DataFrame(mat,index=['Mon','Tues','Wed','Thurs','Fri'], columns=['Almond butter','Peanut butter','Cashew butter'])
Sales['total_sales($)']= total_sales
Sales
```
